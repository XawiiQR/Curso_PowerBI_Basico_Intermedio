# **<font color="#FFFFFF"> ¿Qué es Business Intelligence?</font>**
</td>
</tr>
</table>

**Business Intelligence (BI)** se refiere al conjunto de estrategias, tecnologías y procesos que permiten a las organizaciones recopilar, analizar y presentar datos comerciales.

El objetivo principal de BI es ayudar a las empresas a tomar decisiones informadas basadas en datos.

<table width='99%'>
<tr>
<td bgcolor='#FFBA39'>

## **<font color="#000000"> Componentes claves de la Inteligencia de Negocios</font>**
</td>
</tr>
</table>

>* **Recopilación de Datos**: Involucra la obtención de datos de diversas fuentes, como bases de datos, hojas de cálculo y sistemas ERP.
>* **Almacenamiento de Datos**: Los datos recopilados se almacenan en data warehouses o bases de datos, donde se estructuran para facilitar el análisis.
>* **Análisis de Datos**: Se utilizan herramientas y técnicas para analizar los datos, identificar patrones, tendencias y obtener información valiosa.
>* **Visualización de Datos**: Los resultados del análisis se presentan a través de dashboards, gráficos y reportes, facilitando la comprensión de la información.
>* **Toma de Decisiones**: La información obtenida permite a los directivos y equipos tomar decisiones estratégicas basadas en evidencias y datos concretos.

# **<font color="#FFFFFF"> ¿Qué es Power BI?</font>**
</td>
</tr>
</table>

**Power BI** es una herramienta de Business Intelligence (BI) desarrollada por Microsoft que permite a los usuarios transformar datos en información visual interactiva y comprensible. Su objetivo es facilitar la toma de decisiones informadas mediante el análisis y la visualización de datos.

<table width='99%'>
<tr>
<td bgcolor='#FFBA39'>

## **<font color="#000000"> Características</font>**
</td>
</tr>
</table>

>* **Recopilación de Datos**: Power BI puede conectarse a diversas fuentes de datos, como bases de datos, archivos Excel, servicios en la nube y APIs.
>* **Modelado de Datos**: Permite limpiar, transformar y modelar datos antes de visualizarlos. Los usuarios pueden crear relaciones entre diferentes conjuntos de datos.
>* **Visualización**: Ofrece una amplia variedad de opciones de visualización, como gráficos, tablas, mapas y dashboards personalizados.
>* **Informes Interactivos**: Los usuarios pueden crear informes interactivos que permiten explorar datos a través de filtros y segmentaciones.
>* **Colaboración**: Los informes y dashboards pueden compartirse fácilmente dentro de la organización, lo que facilita la colaboración entre equipos.
>* **Acceso en la Nube**: Power BI ofrece una versión en la nube que permite acceder a los informes desde cualquier dispositivo con conexión a internet.
>* **Integración con Otras Herramientas**: Se integra bien con otras aplicaciones de Microsoft, como Excel y Azure, así como con herramientas de terceros.

# **<font color="#FFFFFF"> Componentes de Power BI: Power BI Desktop, Power BI Services, Power BI Mobile</font>**
</td>
</tr>
</table>

>* **Power BI Desktop**: Es una aplicación de escritorio que permite crear informes y visualizaciones de datos.
>* **Power BI Service**: Es la plataforma en la nube que permite la publicación, compartición y colaboración de informes y dashboards.
>* **Power BI Mobile**: Aplicación móvil que permite acceder a informes y dashboards desde dispositivos móviles (smartphones y tablets).
